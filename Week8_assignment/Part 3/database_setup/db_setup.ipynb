# Part 3 - Database Setup

Loads the cleaned CSV files from **Part 2/cleaned_data** into a SQLite database located in **Part 3/database**.

```python

from pathlib import Path
import sqlite3
import pandas as pd

# ==========================
# Paths
# ==========================
CLEAN_DATA = Path("../../Part 2/cleaned_data")

DATABASE_DIR = Path("../database")
DATABASE_DIR.mkdir(parents=True, exist_ok=True)

DB_PATH = DATABASE_DIR / "ecommerce.db"

# ==========================
# Functions
# ==========================
def build_database():
    if DB_PATH.exists():
        DB_PATH.unlink()

    conn = sqlite3.connect(DB_PATH)

    customers = pd.read_csv(CLEAN_DATA / "customers_clean.csv")
    products = pd.read_csv(CLEAN_DATA / "products_clean.csv")
    orders = pd.read_csv(CLEAN_DATA / "orders_clean.csv")
    order_items = pd.read_csv(CLEAN_DATA / "order_items_clean.csv")

    customers.to_sql("customers", conn, if_exists="replace", index=False)
    products.to_sql("products", conn, if_exists="replace", index=False)
    orders.to_sql("orders", conn, if_exists="replace", index=False)
    order_items.to_sql("order_items", conn, if_exists="replace", index=False)

    return conn

def run_query(conn, sql):
    return pd.read_sql_query(sql, conn)

# ==========================
# Build Database
# ==========================
conn = build_database()

print("="*50)
print("Database Created Successfully")
print("="*50)
print(f"Database Location : {DB_PATH.resolve()}")

for table in ["customers","products","orders","order_items"]:
    rows = conn.execute(f"SELECT COUNT(*) FROM {table}").fetchone()[0]
    print(f" {table:<12}: {rows:,} rows")

print("\nDatabase is ready for SQL analysis.")
```
